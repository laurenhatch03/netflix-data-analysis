```python
import pandas as pd
import numpy as np
from google.colab import drive

# Connecting Google Drive
drive.mount('/content/drive')

# file path to googel drive
file_path = '/content/drive/MyDrive/Netflix/all-weeks-global.xlsx'
```

```python
# reading the raw file
rawtop10 = pd.read_excel(file_path)
rawtop10.head()
```

```python
# dropping duplicates
top10 = rawtop10.drop_duplicates()

# converting to datetime for time-based extraction
top10['week'] = pd.to_datetime(top10['week'])

#removes record after 2025
top10 = top10[top10['week'].dt.year < 2026]

# extracting year and month components for later use
top10["year"] = top10["week"].dt.year
top10["month"] = top10["week"].dt.month
```

```python
# checking datatypes for show titles
top10["show_title"].map(type).value_counts()
```

```python
# checking data types for season titles
# (floats are ok because of the missing values)
top10["season_title"].map(type).value_counts()
```

```python
# uploading to google drive
file_path = '/content/drive/MyDrive/Netflix/Top10.csv'
top10.to_csv(file_path, index=False)
```
